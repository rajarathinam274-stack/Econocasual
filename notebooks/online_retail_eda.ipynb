# Online Retail exploratory analysis

This notebook summarizes **observed positive-sales transactions** using the reusable EconoCausal preprocessing and EDA functions. It does not identify causal price effects.

Run `python scripts/download_dataset.py` from the repository root first if `data/raw` is empty.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from econocausal.preprocessing import preprocess_transactions
from econocausal.eda import plot_eda

raw_dir = ROOT / 'data' / 'raw'
candidates = sorted([*raw_dir.glob('*.csv'), *raw_dir.glob('*.xlsx'), *raw_dir.glob('*.xls')])
if not candidates:
    raise FileNotFoundError('No extracted dataset found. Run: python scripts/download_dataset.py')
source_path = candidates[0]
source_path
```

```python
if source_path.suffix.lower() == '.csv':
    raw = pd.read_csv(source_path)
else:
    raw = pd.read_excel(source_path)
transactions = preprocess_transactions(raw)
transactions.head()
```

```python
output_dir = ROOT / 'outputs' / 'eda'
summary = plot_eda(transactions, output_dir, top_n=10)
{
    'rows_after_preprocessing': len(transactions),
    'observed_positive_sales_revenue': summary['positive_sales_revenue'],
    'observed_positive_sales_units': summary['positive_sales_units'],
    'output_dir': summary['output_dir'],
    'interpretation_warning': summary['interpretation_warning'],
}
```
